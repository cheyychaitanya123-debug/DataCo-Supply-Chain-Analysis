```python
category_profit = (
    df.dropna(subset=["Order Profit Per Order"])
      .groupby("Category Name")
      .agg(
          Total_Profit=("Order Profit Per Order", "sum"),
          Order_Items=("Order Item Id", "nunique")
      )
      .sort_values("Total_Profit", ascending=False)
      .round(2)
)

category_profit
```

```python
category_profit["Avg_Profit_Per_Item"] = (
    category_profit["Total_Profit"] /
    category_profit["Order_Items"]
).round(2)

category_profit.sort_values(
    "Avg_Profit_Per_Item",
    ascending=False
)
```

```python
category_profitability = (
    df.dropna(subset=["Order Profit Per Order"])
      .groupby("Category Name")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Order Profit Per Order", "sum"),
          Order_Items=("Order Item Id", "nunique")
      )
)

category_profitability["Profit_Margin_%"] = (
    category_profitability["Total_Profit"]
    / category_profitability["Total_Sales"]
    * 100
).round(2)

category_profitability = category_profitability.sort_values(
    "Total_Profit",
    ascending=False
)

category_profitability
```

```python
sales_median = category_profitability["Total_Sales"].median()
margin_median = category_profitability["Profit_Margin_%"].median()

high_sales_low_margin = category_profitability[
    (category_profitability["Total_Sales"] >= sales_median) &
    (category_profitability["Profit_Margin_%"] < margin_median)
].sort_values("Total_Sales", ascending=False)

high_sales_low_margin
```

```python
category_profit_ratio = (
    df.groupby("Category Name")
      .agg(
          Avg_Profit_Ratio=("Order Item Profit Ratio", "mean")
      )
)

category_profit_ratio["Avg_Profit_Ratio_%"] = (
    category_profit_ratio["Avg_Profit_Ratio"] * 100
).round(2)

category_profit_ratio = category_profit_ratio.drop(
    columns="Avg_Profit_Ratio"
)

category_profit_ratio.sort_values(
    "Avg_Profit_Ratio_%",
    ascending=True
)
```

```python
weighted_profit_ratio = (
    df.groupby("Category Name")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Order Profit Per Order", "sum")
      )
)

weighted_profit_ratio["Weighted_Profit_Ratio_%"] = (
    weighted_profit_ratio["Total_Profit"]
    / weighted_profit_ratio["Total_Sales"]
    * 100
).round(2)

weighted_profit_ratio.sort_values(
    "Weighted_Profit_Ratio_%",
    ascending=True
)
```

The real business insight is:

High sales do not automatically translate into high profit margin. In this dataset, several of the largest categories generate substantial absolute profit, but their profit margin remains around 10–11%. Discount rates are also very similar across categories, so discounting does not explain the difference.

```python
customer_sales = (
    df.groupby("Customer Id")
      .agg(
          Total_Sales=("Sales", "sum"),
          Orders=("Order Id", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

customer_sales.head(10)
```

```python
top_10_customer_sales = customer_sales.head(10)["Total_Sales"].sum()

total_customer_sales = customer_sales["Total_Sales"].sum()

top_10_share = (
    top_10_customer_sales /
    total_customer_sales *
    100
)

top_10_share
```

That means:

The top 10 customers account for approximately 25.04% of total sales.

```python
customer_sales_sorted = customer_sales.sort_values(
    "Total_Sales",
    ascending=False
)

total_sales = customer_sales_sorted["Total_Sales"].sum()
total_customers = len(customer_sales_sorted)

for percentage in [1, 5, 10]:
    n = int(total_customers * percentage / 100)
    
    sales_share = (
        customer_sales_sorted.head(n)["Total_Sales"].sum()
        / total_sales
        * 100
    )
    
    print(f"Top {percentage}% of customers: {sales_share:.2f}% of sales")
```

```python
#How many order items do the top customers actually have?
top_customers = customer_sales.head(10).index

df[df["Customer Id"].isin(top_customers)].groupby("Customer Id").agg(
    Order_Items=("Order Item Id", "nunique"),
    Orders=("Order Id", "nunique"),
    Sales=("Sales", "sum")
).sort_values("Sales", ascending=False)
```

```python
order_value = (
    df.groupby("Order Id")
      .agg(
          Items=("Order Item Id", "nunique"),
          Sales=("Sales", "sum"),
          Profit=("Order Profit Per Order", "sum")
      )
)

order_value.head()
```

```python
order_size_analysis = (
    order_value.groupby("Items")
    .agg(
        Orders=("Sales", "count"),
        Avg_Sales=("Sales", "mean"),
        Avg_Profit=("Profit", "mean")
    )
    .round(2)
)

order_size_analysis
```

```python
order_size_analysis["Profit_Per_Item"] = (
    order_size_analysis["Avg_Profit"]
    / order_size_analysis.index
).round(2)

order_size_analysis
```

Business conclusion

Larger orders generate higher sales and total profit because they contain more items, while profit per item remains relatively stable for 2–5 item orders.

```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect("../SQL/dataco_supply_chain.db")

# Load CSV
df = pd.read_csv(
    "../Data/raw/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

# Upload dataframe to SQLite
df.to_sql(
    "supply_chain",
    conn,
    if_exists="replace",
    index=False
)

print("Data loaded successfully!")
```
